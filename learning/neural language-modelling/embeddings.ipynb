
$$\text{Token}\rightarrow\text{ID}\rightarrow\text{One-hot}\rightarrow\text{Embedding}$$

$$\\x_i^\top E=E[i,:]$$

```python
import numpy as np
```

Tokenisation

```python
vocab = ["cat", "dog", "puppy", "runs", "sleeps"]
vocab_dict = {word: index for index, word in enumerate(vocab)}

print(vocab_dict)
```

One Hot Encoding

```python
vocab_size = len(vocab)
matrix = np.eye(vocab_size, dtype=float)

print(matrix)
```

Embedding Matrix

```python
E = np.array([
    [ 0.2, -0.1,  0.5],
    [ 0.7,  0.3, -0.2],
    [ 0.6,  0.4, -0.1],
    [-0.3,  0.8,  0.2],
    [ 0.1, -0.5,  0.9]
], dtype=float)

print(E)
```

Lookup Methods

```python
embeddings_A = matrix @ E

all_token_ids = np.array([vocab_dict[word] for word in vocab])

embeddings_B = E[all_token_ids]

print(f"Embeddings A shape: {embeddings_A.shape}")
print(f"Embeddings B shape: {embeddings_B.shape}\n")

print(f"Both Equal? {np.allclose(embeddings_A, embeddings_B)} \n")

print("Word Embeddings Breakdown:")
for word in vocab:
    token_id = vocab_dict[word]
    print(f"'{word}' (ID: {token_id}) -> {embeddings_A[token_id]}")
```
